##**Data processing**

**Check missing values**

```python
#Statistics of the number of missing values for each column in the dataset
check_missing_val = AAPL_data.isnull().sum()
print("Statistics of the number of missing values for each column in the dataset")
print(check_missing_val)
```

**Data normalization**

```python
scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(AAPL_data[['Open', 'High', 'Low', 'Close', 'Adj Close', 'Volume']])
```

**Create sequence data**

```python
def create_multivariate_sequences(data, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])          #Lấy chuỗi con với nhiều đặc trưng
        y.append(data[i+seq_length][0])         #Dự đoán cột "Open" tiếp theo
    return np.array(X), np.array(y)

seq_length = 60
X, y = create_multivariate_sequences(scaled_data, seq_length)
```

```python
print(X.shape) #(samples, seq_length, features)
print(y.shape)
```

**Split dataset**

```python
#80% train, 20% test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

print("Train shape:", X_train.shape, y_train.shape)
print("Test shape:", X_test.shape, y_test.shape)
```

##**Model**

```python
class SimpleRNN(nn.Module):
  def __init__(self, input_size, hidden_size, num_layers, output_size):
    super(SimpleRNN, self).__init__()
    self.hidden_size = hidden_size
    self.num_layers = num_layers

    self.rnn = nn.RNN(input_size, hidden_size, num_layers, batch_first=True, nonlinearity='tanh')

    self.fc = nn.Linear(hidden_size, output_size)

  def forward(self, x):
    #Create hidden state full 0
    h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size).to(x.device)

    out, _ = self.rnn(x, h0) #_ is the last hidden state so we can skip

    out = self.fc(out[:, -1, :])
    return out

input_size = X.shape[2]
hidden_size = 64
num_layers = 2
output_size = 1

rnn_model = SimpleRNN(input_size, hidden_size, num_layers, output_size)
print(rnn_model)
```

##**Training**

```python
#Convert "numpy" to "tensor"
X_train_t = torch.tensor(X_train, dtype=torch.float32)
y_train_t = torch.tensor(y_train, dtype=torch.float32).view(-1, 1)
X_test_t = torch.tensor(X_test, dtype=torch.float32)
y_test_t = torch.tensor(y_test, dtype=torch.float32).view(-1, 1)
```

```python
#Loss and Optimizer
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(rnn_model.parameters(), lr=0.001)
```

```python
#Training
losses = []
epochs = 50
for epoch in range(epochs):
    rnn_model.train()
    output = rnn_model(X_train_t)
    loss = criterion(output, y_train_t)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    losses.append(loss.item())

    print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")
```

```python
#Plot Loss
plt.figure(figsize=(10, 5))
plt.plot(losses, label='Training Loss', color='orange', linewidth=2)
plt.title('Training Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Redict
rnn_model.eval()
with torch.no_grad():
  y_pred_scaled = rnn_model(X_test_t).cpu().numpy()

#Inverse transform
y_pred_full = np.zeros((y_pred_scaled.shape[0], scaled_data.shape[1]))
y_pred_full[:, 0] = y_pred_scaled.flatten()
y_pred_inverse = scaler.inverse_transform(y_pred_full)[:, 0]

# Inverse transform for y_test
y_test_full = np.zeros((y_test.shape[0], scaled_data.shape[1]))
y_test_full[:, 0] = y_test.flatten()
y_test_inverse = scaler.inverse_transform(y_test_full)[:, 0]
```

## **Plot**

```python
test_dates = AAPL_data['Date'][-len(y_test_inverse):].reset_index(drop=True)
```

```python
plt.figure(figsize=(20,6))
plt.plot(test_dates, y_test_inverse, label='Actual Price', color='b')
plt.plot(test_dates, y_pred_inverse, label='Predicted Price', color='r')
plt.title('AAPL Opening Price Prediction (RNN)')
plt.xlabel('Time')
plt.ylabel('Price')

plt.gca().xaxis.set_major_locator(mdates.YearLocator())
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.show()
```
